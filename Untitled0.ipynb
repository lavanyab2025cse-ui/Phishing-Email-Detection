```python
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, confusion_matrix, ConfusionMatrixDisplay

# -------------------------------
# DATASET
# -------------------------------

data = {
    "text": [
            "Hello, your order has been shipped and will arrive tomorrow.",
                    "Your monthly bank statement is ready. Please log in to your account.",
                            "Meeting is scheduled for tomorrow at 10 AM. Please attend.",
                                    "Thank you for your purchase. Your receipt is attached.",
                                            "Your application has been received successfully.",
                                                    "Please find the assignment attached for your review.",
                                                            "Your appointment is confirmed for Monday.",
                                                                    "Your electricity bill is available in your account.",
                                                                            "Welcome to our newsletter. Here are this month's updates.",
                                                                                    "Your course registration has been completed successfully.",

                                                                                            "Urgent! Your account has been suspended. Click here immediately to verify.",
                                                                                                    "Congratulations! You won a prize. Click this link to claim your reward.",
                                                                                                            "Your bank account will be closed today. Verify your password now.",
                                                                                                                    "Important! Click the link below to confirm your account details.",
                                                                                                                            "You have won a lottery. Send your details to receive the money.",
                                                                                                                                    "Your password has expired. Click here to reset it immediately.",
                                                                                                                                            "Security alert! Confirm your username and password using this link.",
                                                                                                                                                    "Claim your free gift now by clicking this suspicious link.",
                                                                                                                                                            "Your account has been selected for a special reward. Act now.",
                                                                                                                                                                    "Urgent verification required. Enter your login details to avoid suspension."
                                                                                                                                                                        ],

                                                                                                                                                                            "label": [
                                                                                                                                                                                    "Safe", "Safe", "Safe", "Safe", "Safe",
                                                                                                                                                                                            "Safe", "Safe", "Safe", "Safe", "Safe",

                                                                                                                                                                                                    "Phishing", "Phishing", "Phishing", "Phishing", "Phishing",
                                                                                                                                                                                                            "Phishing", "Phishing", "Phishing", "Phishing", "Phishing"
                                                                                                                                                                                                                ]
                                                                                                                                                                                                                }

                                                                                                                                                                                                                df = pd.DataFrame(data)

                                                                                                                                                                                                                print("Dataset loaded successfully!")
                                                                                                                                                                                                                print("Total emails:", len(df))

                                                                                                                                                                                                                # -------------------------------
                                                                                                                                                                                                                # SPLIT DATA
                                                                                                                                                                                                                # -------------------------------

                                                                                                                                                                                                                X_train, X_test, y_train, y_test = train_test_split(
                                                                                                                                                                                                                    df["text"],
                                                                                                                                                                                                                        df["label"],
                                                                                                                                                                                                                            test_size=0.3,
                                                                                                                                                                                                                                random_state=42,
                                                                                                                                                                                                                                    stratify=df["label"]
                                                                                                                                                                                                                                    )

                                                                                                                                                                                                                                    # -------------------------------
                                                                                                                                                                                                                                    # TEXT FEATURE EXTRACTION
                                                                                                                                                                                                                                    # -------------------------------

                                                                                                                                                                                                                                    vectorizer = TfidfVectorizer()

                                                                                                                                                                                                                                    X_train_vectorized = vectorizer.fit_transform(X_train)
                                                                                                                                                                                                                                    X_test_vectorized = vectorizer.transform(X_test)

                                                                                                                                                                                                                                    # -------------------------------
                                                                                                                                                                                                                                    # TRAIN MODEL
                                                                                                                                                                                                                                    # -------------------------------

                                                                                                                                                                                                                                    model = LogisticRegression()
                                                                                                                                                                                                                                    model.fit(X_train_vectorized, y_train)

                                                                                                                                                                                                                                    # -------------------------------
                                                                                                                                                                                                                                    # PREDICTION
                                                                                                                                                                                                                                    # -------------------------------

                                                                                                                                                                                                                                    y_pred = model.predict(X_test_vectorized)

                                                                                                                                                                                                                                    # -------------------------------
                                                                                                                                                                                                                                    # ACCURACY
                                                                                                                                                                                                                                    # -------------------------------

                                                                                                                                                                                                                                    accuracy = accuracy_score(y_test, y_pred)

                                                                                                                                                                                                                                    print("\n==============================")
                                                                                                                                                                                                                                    print("PHISHING EMAIL DETECTION")
                                                                                                                                                                                                                                    print("==============================")

                                                                                                                                                                                                                                    print("\nAccuracy:", round(accuracy * 100, 2), "%")

                                                                                                                                                                                                                                    # -------------------------------
                                                                                                                                                                                                                                    # CONFUSION MATRIX
                                                                                                                                                                                                                                    # -------------------------------

                                                                                                                                                                                                                                    cm = confusion_matrix(
                                                                                                                                                                                                                                        y_test,
                                                                                                                                                                                                                                            y_pred,
                                                                                                                                                                                                                                                labels=["Safe", "Phishing"]
                                                                                                                                                                                                                                                )

                                                                                                                                                                                                                                                print("\nConfusion Matrix:")
                                                                                                                                                                                                                                                print(cm)

                                                                                                                                                                                                                                                disp = ConfusionMatrixDisplay(
                                                                                                                                                                                                                                                    confusion_matrix=cm,
                                                                                                                                                                                                                                                        display_labels=["Safe", "Phishing"]
                                                                                                                                                                                                                                                        )

                                                                                                                                                                                                                                                        disp.plot()
                                                                                                                                                                                                                                                        plt.title("Phishing Email Detection - Confusion Matrix")
                                                                                                                                                                                                                                                        plt.show()

                                                                                                                                                                                                                                                        # -------------------------------
                                                                                                                                                                                                                                                        # TEST A NEW EMAIL
                                                                                                                                                                                                                                                        # -------------------------------

                                                                                                                                                                                                                                                        new_email = [
                                                                                                                                                                                                                                                            "Urgent! Click this link to verify your password and account immediately."
                                                                                                                                                                                                                                                            ]

                                                                                                                                                                                                                                                            new_email_vectorized = vectorizer.transform(new_email)

                                                                                                                                                                                                                                                            prediction = model.predict(new_email_vectorized)

                                                                                                                                                                                                                                                            print("\nTest Email:")
                                                                                                                                                                                                                                                            print(new_email[0])

                                                                                                                                                                                                                                                            print("\nPrediction:", prediction[0])

                                                                                                                                                                                                                                                            # -------------------------------
                                                                                                                                                                                                                                                            # CREATE REPORT
                                                                                                                                                                                                                                                            # -------------------------------

                                                                                                                                                                                                                                                            report = f"""
                                                                                                                                                                                                                                                            PHISHING EMAIL DETECTION MODEL
                                                                                                                                                                                                                                                            ==============================

                                                                                                                                                                                                                                                            Model: Logistic Regression
                                                                                                                                                                                                                                                            Feature Extraction: TF-IDF

                                                                                                                                                                                                                                                            Total Emails: {len(df)}

                                                                                                                                                                                                                                                            Accuracy: {accuracy * 100:.2f}%

                                                                                                                                                                                                                                                            Confusion Matrix:
                                                                                                                                                                                                                                                            {cm}

                                                                                                                                                                                                                                                            Test Email:
                                                                                                                                                                                                                                                            {new_email[0]}

                                                                                                                                                                                                                                                            Prediction:
                                                                                                                                                                                                                                                            {prediction[0]}

                                                                                                                                                                                                                                                            The model classifies emails as either Safe or Phishing
                                                                                                                                                                                                                                                            based on their textual content and keywords.
                                                                                                                                                                                                                                                            """

                                                                                                                                                                                                                                                            with open("phishing_detection_report.txt", "w") as file:
                                                                                                                                                                                                                                                                file.write(report)
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, confusion_matrix

emails=[
"Your account has been suspended. Click here immediately.",
"Congratulations! You won a prize. Click this link to claim.",
"Your bank account will be closed today. Verify your password.",
"Important! Click the link below to confirm your account details.",
"You have won a lottery. Send your details to receive the money.",
"Your password has expired. Click here to reset it immediately.",
"Security alert! Confirm your username and password using this link.",
"Claim your free gift now by clicking this suspicious link.",
"Urgent verification required. Enter your login details.",
"Please find the assignment attached for your review.",
"Your appointment is confirmed for Monday.",
"Your electricity bill is available in your account.",
"Welcome to our newsletter. Here are this month's updates.",
"Your course registration has been completed successfully.",
"Your project meeting is scheduled for tomorrow.",
"Thank you for submitting your assignment."
]

labels=[
"phishing",
"phishing",
"phishing",
"phishing",
"phishing",
"phishing",
"phishing",
"phishing",
"phishing",
"safe",
"safe",
"safe",
"safe",
"safe",
"safe",
"safe"
]

vectorizer=TfidfVectorizer()
X=vectorizer.fit_transform(emails)

X_train,X_test,y_train,y_test=train_test_split(X,labels,test_size=0.25,random_state=42)

model=LogisticRegression()
model.fit(X_train,y_train)

predictions=model.predict(X_test)

accuracy=accuracy_score(y_test,predictions)

print("PHISHING EMAIL DETECTION MODEL")
print("--------------------------------")
print("Accuracy:",accuracy)

print("\nCONFUSION MATRIX")
print(confusion_matrix(y_test,predictions))

email=input("\nEnter an email message: ")

email_vector=vectorizer.transform([email])
result=model.predict(email_vector)

print("\nPrediction:",result[0])

if result[0]=="phishing":
    print("Warning: This email may be a phishing email.")
else:
    print("This email appears to be safe.")
```
